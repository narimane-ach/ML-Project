```python
import seaborn as sns
import matplotlib as plt
import matplotlib.pyplot as plt  
```

```python
df = pd.read_csv("nail_colors_simple.csv", sep=";")

```

```python
df["r"] = df["hex_code"].str[1:3].apply(lambda x: int(x, 16))
df["g"] = df["hex_code"].str[3:5].apply(lambda x: int(x, 16))
df["b"] = df["hex_code"].str[5:7].apply(lambda x: int(x, 16))
```

```python
corr = df[["r", "g", "b"]].corr()
print(corr)
```

```python

sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="color_family",order=df["color_family"].value_counts().index)
plt.title("Number of nail colors per family")
plt.xlabel("Color family")
plt.ylabel("Number of colors")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
df = pd.read_csv("nail_colors_simple.csv", sep=";")

plt.figure(figsize=(14, 7))                     
sns.scatterplot(data=df, x="color_name", y="color_family", s=100)

plt.xticks(rotation=60, ha="right", fontsize=10) 
plt.yticks(fontsize=10)
plt.xlabel("Color name", fontsize=12)
plt.ylabel("Color family", fontsize=12)
plt.title("Nail colors by family", fontsize=14)
plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()                                
plt.show()
```
